# SpaceX Falcon 9 — End-to-end analysis
Historical IBM capstone dataset; see README for provenance and caveats.

```
import pandas as pd, sqlite3
from pathlib import Path
df = pd.read_csv('data.csv')
print(df.shape,df.Class.value_counts().to_dict())
display(df.head())
```

```
display(df.groupby('LaunchSite').agg(launches=('Class','size'), positive=('Class','sum'), success_pct=('Class',lambda x:round(x.mean()*100,1))))
```

```
import matplotlib.pyplot as plt
df.assign(Year=pd.to_datetime(df.Date).dt.year).groupby('Year').Class.agg(['count','sum']).plot(kind='bar');plt.show()
```

```
con=sqlite3.connect(':memory:');df.to_sql('launches',con,index=False)
display(pd.read_sql_query('SELECT LaunchSite, COUNT(*) launches, SUM(Class) successes FROM launches GROUP BY LaunchSite',con))
```
